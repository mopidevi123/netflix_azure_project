```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
BRONZE_PATH = "abfss://bronze@pravdatalake.dfs.core.windows.net"
```

```python
titles_df = spark.read.format("delta").load(
    f"{BRONZE_PATH}/netflix_titles"
)

directors_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/netflix_directors"
)

countries_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/netflix_countries"
)

category_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/netflix_category"
)

cast_df = spark.read.format("csv").load(
    f"{BRONZE_PATH}/netflix_cast"
)
```

```python
print("Titles:", titles_df.count())
print("Directors:", directors_df.count())
print("Countries:", countries_df.count())
print("Categories:", category_df.count())
print("Cast:", cast_df.count())

titles_df.printSchema()
```
